```python
from sklearn import preprocessing
import numpy as np
```

**sklearn.preprocessing:** Un módulo de la biblioteca de Python Sklearn que ofrece diversas funciones y clases de transformacion para procesar y limpiar datos antes de usarlos en modelos de Machine Learning.

**Numpy:** Una función de Python desarrollada en C que me permite realizar operaciones matematicas complejas con estructuras de datos como matrices, vectores.

```python
X_train = np.array([[1., -1., 2.],
                   [2., 0., 0.],
                   [0., 1., -1.]])

print(X_train)
```

**array:** Función que se encarga de crear un vector o una matriz de datos en Python.

```python
standarizar = preprocessing.StandardScaler()

X_train_std = standarizar.fit_transform(X_train)
print(X_train_std)
```

**StandardScaler:** Función que me permite estandarizar los datos, ajustando sus valores para que tengan una media de 0 y una desviación estándar de 1.

**fit:** Función que aprende de los datos. y que permite analizar los datos y realizar su calculo respectivo.

**transform:** Función que me permite aplicar la transformación a los datos utilizando los valores calculados por fit y ajustandolos.




```python
X_train_2 = np.array ([[1., -1., 2.],
                      [2., 0., 0.],
                      [0., 1., -1.]])
```

```python
f= preprocessing.StandardScaler()
f=f.fit(X_train)
X_train_2_std = f.transform(X_train_2)
print(X_train_2_std)
```

```python
print(np.array_equiv(X_train_std, X_train_2_std))
```

**equiv:** Función que me permite comprobar si dos matrices o vectores son iguales.
