```python
from sklearn import preprocessing
import numpy as np
```

**sklearn.preprocessing:** Un módulo de la biblioteca de Python Sklearn que ofrece diversas funciones y clases de transformacion para procesar y limpiar datos antes de usarlos en modelos de Machine Learning.

**Numpy:** Una función de Python desarrollada en C que me permite realizar operaciones matematicas complejas con estructuras de datos como matrices, vectores.

```python
# Carga de datos.
X_train = np.array([[1., -4., 5.],
                    [5., 0., 2.],
                    [0., 6., -3.]])
print(X_train)
```

**array:** Función que se encarga de crear un vector o una matriz de datos en Python.

```python
# Limpieza de datos: normalización.
normalizer= preprocessing.MinMaxScaler()
X_train_norm = normalizer.fit_transform(X_train)
```

**MinMaxScaler:** Función que me permite encontrar los valores mínimos y máximos de un conjunton de datos.

**fit:** Función que aprende de los datos. y que permite analizar los datos y realizar su calculo respectivo.

**transform:** Función que me permite aplicar la transformación a los datos  utilizando los valores calculados por fit y ajustandolos.

```python
# Salida de resultados.
print(X_train_norm)
```

```python
# Ejemplo para ver que los resultados son iguales si aplicamos los métodos "fit" y "transform" por separado

X_train_2 = np.copy(X_train)

# Crea un objeto "fitted_normalizer" llamando al método "fit"

f = normalizer.fit(X_train)

# Aplica el objeto sobre los datos "X_train_2" para obtener "X_train_norm_2"

X_train_norm_2= f.transform(X_train)

# Comprueba si "X_train_norm" y "X_train_norm_2" son iguales
print(np.array_equiv(X_train_norm, X_train_norm_2))
```

**copy:** Función que me permite copiar un vector o matriz de una variable.

**equiv:** Función que me permite comprobar si dos matrices o vectores son iguales. 

```python
# Normalizar un nuevo conjunto de datos de test (SOLO HAY QUE APLICAR EL MÉTODO "transform", NO EL "fit")

X_test= np.array([[1., -3., 2.],
                 [3., 0., 0.],
                 [0., 4., -1.]])

X_test_norm = f.transform(X_test)
```
